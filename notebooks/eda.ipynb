# Correlation validation

We want to perform a VIF analysis to check for multicollinearity among the features.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
```

```python
features_for_vif = features.drop(columns=[gold_cols.DATE])
features_for_vif = add_constant(features_for_vif)  # add intercept term

vif_values = pd.Series(
    [
        variance_inflation_factor(features_for_vif.values, i)
        for i in range(features_for_vif.shape[1])
    ],
    index=features_for_vif.columns,
)
```

```python
vif_values.sort_values(ascending=False)
```

The online / offline price features (control variables) have high VIF values (around 7), which makes sense as they are likely correlated with each other. For the media features, the VIF values are generally low (mostly below 2), but there are two exceptions: `social_media_transaction_the_hut_group` and `paid_search_awarenessconsideration_google`, which have VIF values around 7.
